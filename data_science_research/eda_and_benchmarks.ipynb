### 1. Ingestion & Exploratory Data Analysis (EDA) on Job Postings Dataset


```python
with open("../backend/app/data/jobs_dataset.json") as f:
    jobs_data = json.load(f)
df_jobs = pd.DataFrame(jobs_data)
print(f"Total Job Records: {len(df_jobs)}")
df_jobs.head(3)
```

### 2. Analysis of Most Demanded Skills Across Tech Roles


```python
all_skills = [s for sublist in df_jobs["required_skills"] for s in sublist]
skill_counts = pd.Series(all_skills).value_counts()
plt.figure(figsize=(10, 4))
skill_counts.head(10).plot(kind="bar", color="#6366f1")
plt.title("Top 10 Most Demanded Skills in Dataset")
plt.xlabel("Skill")
plt.ylabel("Frequency")
plt.show()
```
